## Overview

In our last class module, we discussed Brexit and other applications that use _panel data._  Panel data involves observing multiple units repeatedly over a period of time, giving us two dimensions to consider: 1) units and 2) time. This structure opens the door to many new empricial research designs and methods that utilize both dimensions to obtain more accurate estimates of treatment effects. It’s important to note that, with panel data, we often focus on estimating the average treatment effect on the treated (ATT) rather than the average treatment effect (ATE).

**What are we doing in this homework?**

In this HW, we will analyze the question of what economic impact Brexit had on the UK using annual GDP as an outcome measure. Specifically, we will use a data set containing annual GDP per hours worked in USD of all OECD countries from 1995-2022 to estimate the effect. The quasi-experimental variation we want to use is that we observe outcomes for the unit of interest, the UK, before Brexit happened in 2020 and after. 

In part 1 of this homework, we will again begin with an exploratory data analysis, including calculating summary statistics. In part 2 of this homework, we will make use of the two new tools we learned in class to causally identify parameters with panel data: _difference-in-differences_ and _matrix completion_. In the bonus question, we will analyze the data set using _synthetic control methods_.

1. As we discussed in class, this sort of natural experiment -- where we have some units treated at some time period(s), while other, different, units are not treated -- shows up a lot. In this problem set we will get practice with this sort of problem.

2. Difference-in-difference designs are one of the most widely used quasi-experimental method for panel data.

### Part 2c: Can we do better?

We saw that the three different control groups have their own advantages and disadvantages - some are following more similar trends to the UK while others are also closer in levels to the UK. Naturally, you might wonder if we can do better. As you saw in class, that is exactly where the **synthetic control estimator** comes in. 

Let's implement it and see how it performs!

You might have to run the following cell if you don't have the toolz pacakge installed already.

```python
pip install toolz
```

```python
# SC from scratch
# Note that this is beyond what you need to know!

# Packages
import numpy as np
from scipy.optimize import fmin_slsqp
from toolz import partial

#####################################
#### Helper functions 
#####################################

# Define loss function for optimization procedure
def loss_w(W, X, y) -> float:
    return np.sqrt(np.mean((y - X.dot(W))**2)) # Sum of squared residuals


# Function to get weights
def get_w(X, y):

    """
    This function returns SC weights.

    X: T x N_c matrix of panel data from control units
    y: T x 1 vector of treated unit
    
    """
    
    w_start = [1/X.shape[1]]*X.shape[1] # initial value for the weights

    weights = fmin_slsqp(partial(loss_w, X=X, y=y),
                         np.array(w_start),
                         f_eqcons=lambda x: np.sum(x) - 1, # restrict weights to sum to 1
                         bounds=[(0.0, 1.0)]*len(w_start), # bound weights to be between 0 and 1
                         disp=False)
    return weights

#####################################
#### Analysis 
#####################################

# Get SC weights
T0 = 25 # Time of Brexit
X = np.transpose(gdp.iloc[:29,:T0]) # All GDP values over time pre-Brexit, but the UK's, transposed
y = gdp.iloc[29,:T0] # UK's GDP over time pre-Brexit
uk_weights = get_w(X, y)
print("The calculated weights are:", np.round(uk_weights, 4))
print(" ")
print(" ")

# Which countries' weights are positive
pos_weights = oecd.iloc[np.where(uk_weights > 0)[0],0] # gets country names of countries with positive weight
print("The synthetic control gives positive weights to units:",*pos_weights)
```
